```python
import base64
```

## Image to text

```python
def image_to_base64(image_path):
    with open(image_path, "rb") as image_file:
        encoded_string = base64.b64encode(image_file.read()).decode('utf-8')
    return encoded_string
```

```python
# Demonstrate the function with 'diagram.jpeg'
image_file_path = 'diagram.jpeg'
base64_string = image_to_base64('diagram.jpeg')

if base64_string.startswith("Error"):
    print(base64_string)
else:
    print(f"Base64 encoded string (first 100 characters):\n{base64_string[:100]}...")
    print(f"Total length of base64 string: {len(base64_string)}")
```

```python
output_filename = 'diagram_base64.txt'
with open(output_filename, 'w') as f:
    f.write(base64_string)

print(f"Base64 string saved to '{output_filename}'")
```

## Text back to Image

```python
def base64_to_image(base64_string, output_image_path):
    image_data = base64.b64decode(base64_string)

    with open(output_image_path, 'wb') as f:
        f.write(image_data)
    return f"Image successfully saved to '{output_image_path}'"
```

```python
# Demonstrate the function
output_image_path = 'diagram_restored.jpeg'
result_message = base64_to_image(base64_string, output_image_path)
print(result_message)
```

## How to read special characters using utf-8

```python
sample_string = "Hello, world! This is a test string with some special characters like é, ü, and я."

# Encode the string to UTF-8 bytes
utf8_bytes = sample_string.encode('utf-8')

print(f"Original string: {sample_string}")
print(f"UTF-8 encoded bytes: {utf8_bytes}")
print(f"Type of encoded bytes: {type(utf8_bytes)}")
```
