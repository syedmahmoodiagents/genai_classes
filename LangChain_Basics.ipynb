# Chaining and Parsers

```python
llm = ChatHuggingFace(llm=HuggingFaceEndpoint(repo_id="meta-llama/Llama-3.1-8B-Instruct"))
```

```python
prompt = ChatPromptTemplate.from_template( "Explain {tp} in one sentence, formatted as a JSON object with a single key named 'expl'.")
```

```python
from langchain_core.output_parsers import JsonOutputParser
```

```python
chain = prompt | llm
```

```python
res = chain.invoke({"tp": "quantum ai"})
```

```python
JsonOutputParser().parse(res.content)
```

```python
jsonparser = JsonOutputParser()
```

```python
chain2 = prompt | llm | jsonparser
```

```python
# chain2
```

```python
resp = chain2.invoke({"tp": "quantum ai"})
```
