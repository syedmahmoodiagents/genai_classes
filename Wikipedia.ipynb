```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
```

```python
from langchain_community.tools import WikipediaQueryRun
```

```python
from langchain_community.utilities import WikipediaAPIWrapper
```

```python
wrap = WikipediaAPIWrapper(top_k_results=1, doc_content_chars_max=480)
```

```python
wiki = WikipediaQueryRun(api_wrapper=wrap)
```

```python
wiki_answer = wiki.invoke({"query": "Quantum computer"})
```

```python
print(wiki_answer)
```

```python
import os
os.environ['HF_TOKEN'] = ''
```

```python
llm = ChatHuggingFace(llm=HuggingFaceEndpoint(repo_id="openai/gpt-oss-20b"))
```

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools=[wiki],
    system_prompt="take help of wiki tool to find the information"
)
```

```python
# agent.invoke({"messages": [HumanMessage(content="Hi!")]})
```

```python
response = agent.invoke({
    "messages": [
        {"role": "user", "content": "What is Quatum Computer?"}
    ]
})
```
